```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Завантажуємо набір даних
iris_dataset = load_iris()

print("Ключі iris_dataset:", iris_dataset.keys())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'], iris_dataset['target'], random_state=0
)

print("Розмірність X_train:", X_train.shape)
print("Розмірність X_test:", X_test.shape)
```

```python
# Створюємо DataFrame з навчальних даних
iris_dataframe = pd.DataFrame(X_train, columns=iris_dataset.feature_names)

# Будуємо матрицю розсіювання, забарвлюючи точки за класом y_train
pd.plotting.scatter_matrix(
    iris_dataframe,
    c=y_train,
    figsize=(12, 12),
    marker='o',
    hist_kwds={'bins': 20},
    s=60,
    alpha=.8
)
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Ініціалізуємо модель
knn = KNeighborsClassifier(n_neighbors=1)

# Навчаємо модель на навчальній вибірці
knn.fit(X_train, y_train)
```

```python
# Нові дані у вигляді 2D-масиву NumPy
X_new = np.array([[5, 2.9, 1, 0.2]])

# Прогнозування
prediction = knn.predict(X_new)

print("Номер передбаченого класу:", prediction)
print("Назва виду ірису:", iris_dataset['target_names'][prediction])
```

```python
print("Точність на тестовому наборі: {:.2f}".format(knn.score(X_test, y_test)))
```
